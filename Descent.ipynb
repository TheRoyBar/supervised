```python
import numpy as np
from sklearn.datasets import make_regression
import matplotlib.pyplot as plt
```

```python
x,y = make_regression(n_samples=100,n_features=1,n_informative=1, n_targets=1, noise = 80)
```

```python
plt.scatter(x,y)
```

#using standard sklearn library

```python
from sklearn.linear_model import LinearRegression
```

```python
reg = LinearRegression()
reg.fit(x,y)
```

```python
print(reg.coef_)
print(reg.intercept_)
```

```python
m=28.99
lr=0.1
epochs = 100
b=100
```

```python
for i in range(epochs):
    slopeInstant = -2*np.mean(y - m*x.ravel() - b) 
    b = b-(lr*slopeInstant)

    yPred = m*x+b
    
    plt.plot(x,yPred)
    
plt.scatter(x,y,color= 'blue')
```

```python
plt.scatter(x,y)
plt.plot(x,reg.predict(x),color='red',label='Prediction Line',alpha=0.5)
plt.plot(x,yPred,color='green',label='Descented Line',alpha=0.5)
plt.legend()
plt.show()
```
