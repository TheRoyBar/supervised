```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_regression
```

```python
x,y = make_regression(n_samples=4, n_features=1, n_informative=1, n_targets=1, noise=80, random_state=13)
```

```python
#Using OLS
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
```

```python
lr.fit(x,y)
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
plt.scatter(x,y)
plt.plot(x,lr.predict(x),color = 'red')
```

```python
#Trying out Gradient Descent, m = 78.35 and assuming intercept b = 0
yPred=((78.35*x)+0).reshape(4)
```

```python
yPred
```

```python
plt.scatter(x,y)
plt.plot(x,lr.predict(x),color = 'red',label='OLS')
plt.plot(x,yPred, color = 'blue',label = 'b=0')
plt.legend()
plt.show()
```

```python
m=78.35
b=0

slopeInstant = -2*np.sum(y-m*x.ravel()-b)
slopeInstant
```

```python
learningRate = 0.1

stepSize = slopeInstant*learningRate
stepSize
```

```python
b = b-stepSize
```

```python
b
```

```python
yPred2 = ((78.35*x)+b)

plt.scatter(x,y)
plt.plot(x,lr.predict(x),color = 'red',label='OLS')
plt.plot(x,yPred2, color = 'blue',label = 'b=0')
plt.legend()
plt.show()
```
